# 3️⃣ Dataset Overview

```python
# ============================================================
# LOAD DATASET
# ============================================================

FILE_NAME = "pakwheels_used_cars_clean.csv"

df = pd.read_csv(FILE_NAME)

print("Dataset loaded successfully ✅")

print("\nShape:")
print(df.shape)

print("\nColumns:")
print(df.columns.tolist())

display(df.head())
```

# 4️⃣ Duplicate Check

```python
# ============================================================
# DUPLICATE CHECK
# ============================================================

print("Duplicate rows:", df.duplicated().sum())

df = df.drop_duplicates().reset_index(drop=True)

print("Shape after removing duplicates:", df.shape)
```

# 5️⃣ Missing Values

```python
# ============================================================
# MISSING VALUES
# ============================================================

missing = df.isnull().sum()

missing_percentage = (
    df.isnull().mean() * 100
).round(2)

missing_report = pd.DataFrame({
    "Missing Values": missing,
    "Percentage": missing_percentage
})

display(
    missing_report.sort_values(
        "Missing Values",
        ascending=False
    )
)
```

# 6️⃣ Clean Numeric Columns

```python
# ============================================================
# NUMERIC CLEANING
# ============================================================

numeric_columns = [
    "year",
    "vehicle_age",
    "engine_cc",
    "mileage_km",
    "price",
    "price_million"
]

for col in numeric_columns:

    if col in df.columns:

        df[col] = pd.to_numeric(
            df[col],
            errors="coerce"
        )

print("Numeric columns cleaned ✅")

display(df[numeric_columns].describe())
```

# 7️⃣ Clean Text Columns

```python
# ============================================================
# TEXT CLEANING
# ============================================================

text_columns = [
    "brand",
    "name",
    "fuel_type",
    "transmission"
]

for col in text_columns:

    if col in df.columns:

        df[col] = (
            df[col]
            .astype("string")
            .str.strip()
        )

print("Text columns cleaned ✅")
```

# 9️⃣ Standardize Transmission

```python
# ============================================================
# TRANSMISSION STANDARDIZATION
# ============================================================

if "transmission" in df.columns:

    df["transmission"] = (
        df["transmission"]
        .str.strip()
        .str.title()
    )

display(
    df["transmission"]
    .value_counts()
)
```

# 🔟 Vehicle Age

```python
# ============================================================
# VEHICLE AGE
# ============================================================

CURRENT_YEAR = 2026

df["vehicle_age"] = (
    CURRENT_YEAR - df["year"]
)

df.loc[
    df["vehicle_age"] < 0,
    "vehicle_age"
] = np.nan

print("Vehicle age calculated ✅")

display(
    df[
        ["year", "vehicle_age"]
    ].head(10)
)
```

# 1️⃣1️⃣ Price in Millions

```python
# ============================================================
# PRICE IN MILLIONS
# ============================================================

df["price_million"] = (
    df["price"] / 1_000_000
)

display(
    df[
        ["price", "price_million"]
    ].head(10)
)
```

# 1️⃣2️⃣ Remove Invalid Data

```python
# ============================================================
# INVALID VALUES
# ============================================================

# Invalid year
df.loc[
    ~df["year"].between(
        1950,
        CURRENT_YEAR
    ),
    "year"
] = np.nan


# Invalid price
df.loc[
    df["price"] <= 0,
    "price"
] = np.nan


# Invalid mileage
df.loc[
    df["mileage_km"] < 0,
    "mileage_km"
] = np.nan


# Invalid engine
df.loc[
    df["engine_cc"] < 0,
    "engine_cc"
] = np.nan


print("Invalid values handled ✅")
```

# 1️⃣3️⃣ Missing Value Report Again

```python
# ============================================================
# FINAL MISSING VALUE REPORT
# ============================================================

missing_report = pd.DataFrame({

    "Missing Values":
        df.isnull().sum(),

    "Missing Percentage":
        (
            df.isnull().mean() * 100
        ).round(2)

})

display(
    missing_report.sort_values(
        "Missing Values",
        ascending=False
    )
)
```
